# Enterprise Retail AI Analytics

End-to-end EDA notebook for the portfolio project. **Data is synthetic portfolio data.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

ROOT = Path('..')
df = pd.read_csv(ROOT/'07_DATA/retail_orders_sample.csv', parse_dates=['order_date'])
df.head()
```

## Data quality checks

```python
print('Rows:', len(df))
print('Columns:', len(df.columns))
print('Missing values:', int(df.isna().sum().sum()))
print('Duplicate order IDs:', int(df.order_id.duplicated().sum()))
print(df.order_status.value_counts())
```

## KPI analysis

```python
revenue=df.revenue.sum(); profit=df.profit.sum()
print(f'Revenue: ₹{revenue:,.2f}')
print(f'Profit: ₹{profit:,.2f}')
print(f'Profit margin: {profit/revenue:.2%}')
print(f'Return rate: {(df.order_status=="Returned").mean():.2%}')
print(f'Cancellation rate: {(df.order_status=="Cancelled").mean():.2%}')
```

## Monthly trend

```python
monthly=df.assign(month=df.order_date.dt.to_period('M').astype(str)).groupby('month',as_index=False).agg(revenue=('revenue','sum'),profit=('profit','sum'))
ax=monthly.plot(x='month',y=['revenue','profit'],kind='line',marker='o',figsize=(12,5),title='Monthly Revenue and Profit')
ax.set_ylabel('₹'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Regional performance

```python
region=df.groupby('region',as_index=False).agg(orders=('order_id','count'),revenue=('revenue','sum'),profit=('profit','sum'))
region.sort_values('revenue',ascending=False)
```

## Category profitability

```python
category=df.groupby('category',as_index=False).agg(orders=('order_id','count'),revenue=('revenue','sum'),profit=('profit','sum'))
category['profit_margin']=category['profit']/category['revenue'].replace(0,np.nan)
category.sort_values('profit',ascending=False)
```
